```python
#numpy
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

df["Shipping Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
```

```python
import numpy as np

shipping = df["Shipping Days"].to_numpy()

print("Mean:", np.mean(shipping))
print("Median:", np.median(shipping))
print("Minimum:", np.min(shipping))
print("Maximum:", np.max(shipping))
print("Standard Deviation:", np.std(shipping))
```

```python
total_orders = np.size(shipping)
fast_orders = np.sum(shipping <= 3)
percentage = (fast_orders / total_orders) * 100

print("Total Orders:", total_orders)
print("Orders shipped within 3 days:", fast_orders)
print("Percentage:", percentage)
```
